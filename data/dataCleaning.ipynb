```python
import pandas as pd 
import numpy as np 
import os
```

```python
filename = os.path.join("geopolitical_events.csv")
df = pd.read_csv(filename)
```

```python
df.head(20)
```

```python
#using strftime(Short for string format time) It converts dates into formatted string representations.
df['date'] = pd.to_datetime(df['date'], format='mixed').dt.strftime('%Y-%m-%d')
df.head(20)
```

```python
#checking for null values
df.isna().sum()
```

```python
#checking for inconsitencies?
print(df['event_type'].unique())
print(df['affected_region'].unique())
```

```python
#Check for duplicate event IDs
print("Duplicate IDs:", df['event_id'].duplicated().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
#checking for any outliers
df[['severity', 'duration_days', 'risk_increase']].describe()
```

```python
df.head()
```

```python
df.to_csv('cleaned_geopolitical_events.csv', index=False)
```
